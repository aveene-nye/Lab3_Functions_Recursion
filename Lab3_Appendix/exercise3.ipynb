```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Student Metadata

import functools
import random

LAST_NAME = "ENECILLO"
SEED_NUM = "6"
FAVORITE_ARTIST = "BTS"

execution_logs = []
```

```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Decorator for Logging

def log_diagnostic(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        execution_logs.append(f"Started '{func.__name__}'")
        result = func(*args, **kwargs)
        execution_logs.append(f"Finished '{func.__name__}'")
        return result

    return wrapper
```

```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Generate Telemetry Data

def telemetry_generator():
    random.seed(hash(f"{LAST_NAME}{FAVORITE_ARTIST}{SEED_NUM}"))
    stream = [71.7, "INVALID", 88.5, -14.0, 107.2, None, 67.0]
    for data in stream:
        yield data
```

```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Exception Handling

def validate_telemetry(val):
    if val is None:
        raise ValueError("Missing reading (None)")
    try:
        num = float(val)
    except (ValueError, TypeError):
        raise TypeError(f"'{val}' is not a numeric value")

    if num < 0 or num > 120:
        raise ValueError(f"{num}-C is out of operational range (0-120)")
    
    return num
```

```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Pipeline Analysis and Recursion

def analyze_abnormal_recursion(temp, history=None):
    if history is None:
        history = []

    history.append(round(temp, 1))

    if temp < 80:
        return history

    return analyze_abnormal_recursion(temp - 10, history)

@log_diagnostic
def process_pipeline():
    raw_telemetry = []
    valid_readings = []
    invalid_results = []

    for raw in telemetry_generator():
        raw_telemetry.append(raw)
        try:
            clean_val = validate_telemetry(raw)
            valid_readings.append(clean_val)
        except (ValueError, TypeError) as err:
            invalid_results.append(f"FAIL: {raw} ({err})")

    c_to_f = lambda c: round((c * 9 / 5) + 32, 1)
    transformed_readings = list(map(c_to_f, valid_readings))

    peak_temp = max(valid_readings) if valid_readings else 0
    recursive_trace = analyze_abnormal_recursion(peak_temp)

    if peak_temp >= 100:
        overall_status = "CRITICAL: Overheating detected"
    elif peak_temp >= 85:
        overall_status = "WARNING: Elevated temperature"
    else:
        overall_status = "NORMAL: Fully operational"

    return {
        "raw": raw_telemetry,
        "valid": valid_readings,
        "invalid": invalid_results,
        "transformed": transformed_readings,
        "recursive": recursive_trace,
        "status": overall_status,
    }
```

```python
# [EXERCISE 3 - INTELLIGENT EQUIPMENT MONITORING PIPELINE]
# Execution and Display

execution_logs.clear()
results = process_pipeline()

total_processed = len(results["raw"])
valid_count = len(results["valid"])
invalid_count = len(results["invalid"])

print("=== INTELLIGENT EQUIPMENT MONITORING PIPELINE REPORT ===")
print("Assessment Data:")
print("  Student-Specific Inputs:")
print(f"   LAST_NAME:   {LAST_NAME}")
print(f"   SEED_NUM:   {SEED_NUM}")
print(f"   FAVORITE_ARTIST:   {FAVORITE_ARTIST}")

print(f" Generated Telemetry Data: {results['raw']}")
print(f" Valid/Invalid Results:")
print(f"  - Valid Count:  {valid_count}")
print(f"  - Invalid Count:  {invalid_count}")
for inv in results["invalid"]:
    print(f"   {inv}")

print(f" Processed Results (Fahrenheit): {results['transformed']}")
print(f" Recursive Analysis (Cooling Trace): {results['recursive']}\n")

print("Final Diagnostic Summary:")
print(f"  Total Processed Readings: {total_processed}")
print(f"  Valid Readings:   {valid_count}")
print(f"  Invalid Readings:   {invalid_count}")
print(f"  Detected Abnormal Condition: Peak Temperature {max(results['valid'])}-C")
print(f"  Overall Equipment Status:   {results['status']}\n")

print("Execution Log:")
for log in execution_logs:
    print(f"  - {log}")

print("\nFinal Output:")
print(f"Pipeline complete. Processed {total_processed} items ({valid_count} valid, {invalid_count} invalid). Overall Status: {results['status']}.") 
```
