```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# STUDENT METADATA

import functools
import random

# Enecillo Inputs
LAST_NAME = "ENECILLO"
SEED_NUM = "6"
FAVORITE_ARTIST = "Taylor Swift"

# Decorator logs
execution_logs = []
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Decorator for Logging

def log_diagnostic(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        execution_logs.append(f"Started '{func.__name__}'")
        result = func(*args, **kwargs)
        execution_logs.append(f"Finished '{func.__name__}'")
        return result
    
    return wrapper
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Generating Data

def generate_readings():
    random.seed(hash(f"{LAST_NAME}{FAVORITE_ARTIST}{SEED_NUM}"))

    data = [10, 20, 30, 40, "ERROR", -5, None]
    random.shuffle(data)
    return data
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Core Functions - Validation

@log_diagnostic
def validate_reading(reading):
    if reading is None:
        raise ValueError("Missing value (None)")

    try:
        val = float(reading)
    except (ValueError, TypeError):
        raise TypeError(f"'{reading}' is not a number")
    
    if val < 0 or val > 100:
        raise ValueError(f"{val}-C is out of range (0-100)")

    return val
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Core Functions - Calculation

@log_diagnostic 
def calculate_metrics(valid_list):
    if not valid_list:
        return 0, 0
    avg_temp = round(sum(valid_list) / len(valid_list), 1)
    max_temp = round(max(valid_list), 1)
    return avg_temp, max_temp
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Core Functions - Classification

@log_diagnostic
def classify_condition(avg_temp, max_temp):
    if max_temp >= 90 or avg_temp >= 70:
        return "CRITICAL: Overheating"
    elif max_temp >= 50:
        return "WARNING: High Temp"
    else:
        return "NORMAL: Operational"
```

```python
# [EXERCISE 1 - EQUIPMENT DIAGNOSTIC SYSTEM] 
# Run System

raw_data = generate_readings()

valid_readings = []
validation_results = []

for item in raw_data:
    try:
        clean_val = validate_reading(item) 
        valid_readings.append(clean_val)
        validation_results.append(f"PASS: {item}")
    except (ValueError, TypeError) as err:
        validation_results.append(f"FAIL: {item} ({err})")

avg_temp, max_temp = calculate_metrics(valid_readings)
status = classify_condition(avg_temp, max_temp)

print("=== EQUIPMENT DIAGNOSTIC REPORT ===")
print(f"Generated Data: {raw_data}\n")

print("Validation Results:")
for result in validation_results:
    print(f" {result}")

print(f"\nDiagnostic Results:")
print(f" Average Temperature: {avg_temp}-C")
print(f" Peak Temperature: {max_temp}-C")
print(f" Status: {status}\n")

print("Execution Log:")
for log in execution_logs:
    print(f"  - {log}")

print("\nFinal Output:")
print(f"Equipment status is {status}.")
```
