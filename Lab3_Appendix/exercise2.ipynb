```python
# [EXERCISE 2 - RECURSIVE FAULT TRACE]
# Student Metadata

import functools
import random

LAST_NAME = "ENECILLO"
SEED_NUM = "6"
FAVORITE_ARTIST = "Seventeen"

execution_logs = []
call_count = 0 
```

```python
# [EXERCISE 2 - RECURSIVE FAULT TRACE]
# Decorator for Logging

def log_diagnostic(func):

    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        execution_logs.append(f"Started '{func.__name__}' with argument: {args[0]}")
        result = func(*args, **kwargs)
        execution_logs.append(f"Finished '{func.__name__}")
        return result

    return wrapper
```

```python
# [EXERCISE 2 - RECURSIVE FAULT TRACE]
# Fault Code Generator

def generate_fault_code():
    random.seed(hash(f"{LAST_NAME}{SEED_NUM}{FAVORITE_ARTIST}"))
    return random.randint(500, 1000) 
```

```python
# [EXERCISE 2 - RECURSIVE FAULT TRACE]
# Recursive Trace Function

@log_diagnostic
def trace_fault(code, trace_history):
    global call_count
    call_count += 1

    trace_history.append(code)

    if code <= 9:
        return code

    next_code = code // 2

    return trace_fault(next_code, trace_history)
```

```python
# [EXERCISE 2 - RECURSIVE FAULT TRACE]
# Execution

execution_logs.clear
call_count = 0

initial_fault = generate_fault_code()

trace_history = []
final_result = trace_fault(initial_fault, trace_history)

print("=== RECURSIVE FAULT TRACE ===")
print("Assessment Data:")
print(f" Generated Fault Data: {initial_fault}")
print(f" Recursive Trace: {trace_history}")
print(f" Number of Recursive Calls: {call_count}\n")

print("Execution Log:")
for log in execution_logs:
    print(f" - {log}")

print("\nFinal Output")
print(f" Final resolved condition code: {final_result}")
```
